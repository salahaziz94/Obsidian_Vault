Python's lists are pretty versatile, so we'll be using those. The keys we'll want are the lists' `append` and `pop` method. `append(v)` puts whatever is in the variable `v` at the *end* of the list, while `pop(i)` returns the value at the $i^{\mathrm{th}}$ position in the list (starting from $0$), and deletes it from the list.

```py
a = [1, 2, 3, 4, 5]
a.append(6)
a
```
```
[1, 2, 3, 4, 5, 6]
```

```py
val1 = a.pop(0)
val1
a
```
```
1
[2, 3, 4, 5, 6]
```

Additionally, you can identify the *last* element in the list by setting `i = -1`, so

```py
val2 = a.pop(-1)
val2
a
```
```
6
[2, 3, 4, 5]
```

The only difference between a stack and a queue is that a *stack* always pops the most recently added element, whereas a *queue* always pops the oldest element.
