The next goal is to implement a *depth-first search* function called `dfs`. We will use a search tree library that I wrote for the job. A search tree consists of `SearchNode` objects. They have two properties that we care about. For a `Searchnode` named `node`, the value of `node.label` is the name assigned to that node of the tree, while `node.children` is a list of `SearchNode`s that are the children of `node`. Given a root `SearchNode` called `root` and the label of the `SearchNode` we are trying to reach, called `goal`, you will complete a depth-first search function.

Depth-first search uses a *stack* keep track of the operation. In the first step, a depth-first search adds the *root node* to the stack. Thereafter, it first *pops the node on top of the stack*, then *checks if that node is the goal*, and if so, it exits. If the current node is not the goal node, the algorithm *adds the current node's children to the stack*, and then repeats. If there is nothing left in the stack, it means the tree has been searched completely and the goal has not been found, so the search will exit.

*Your* implementation must return a tuple containing:

1. `True`, if the goal is found in the tree, or `False` if not
2. A list of the *labels of* the current node at each step, in order
3. A list containing lists of the child `SearchNode`s that were added to the stack, at each step

For ease of evaluation, make sure that you add the children of the current node to the list *in order, from first to last*. So, you should append the leftmost or first child first, the second child second, and so on. In general, a depth-first search can add children in any order (the in-class examples add them in the opposite order); we're only requiring this so the evaluator won't have to deal with variability.

```python
from lab1_libs.searchtree import SearchNode

def dfs(root, goal):
    stack = [root]

    current_nodes = [] # this will store the history of current nodes
    added_nodes = [] # this will store the history of nodes that were added

    while stack: # this loop will terminate when the stack is empty
        # pop the last element from the list and assign it to a variable called current

        # your code goes here [1]

        current_nodes.append(current.label)

        # if ever we reach the goal, the function will return early with our history
        if current.label == goal:
            return True, current_nodes, added_nodes # returns a tuple containing our history

        if current.children is not None: # checks if the current node has children
            # add the children to the stack, starting with the first child, in order
            # remember *not* to pop the children; your code will misbehave if you do!

            # your code goes here [2]

            added_nodes.append(current.children)
        else:
            added_nodes.append([]) # to keep track of our steps properly, add an empty list

    # If the search didn't return early, it means we failed (if it did, we'll never reach this point)
    return False, current_nodes, added_nodes
```
